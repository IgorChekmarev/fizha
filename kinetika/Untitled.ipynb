```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

# Список файлов и их названия для легенды
files = [
    ('AION.csv', 'AION'),
    ('BION.csv', 'BION'),
    ('CION.csv', 'CION'),
    ('DION.csv', 'DION'),
    ('EION.csv', 'EION'),
    ('WCC1.csv', 'WCC1'),
    ('WCC2.csv', 'WCC2'),
    ('WCC3.csv', 'WCC3'),
    ('WCC4.csv', 'WCC4'),
    ('WCC5.csv', 'WCC5')
]

# Чтение данных из всех файлов
data_dict = {}
for filename, label in files:
    df = pd.read_csv(filename, skiprows=1, names=['Time_sec', 'Abs'])
    data_dict[label] = df

# Создание графика
plt.figure(figsize=(14, 8))

colors = plt.cm.tab10(np.linspace(0, 1, len(files)))

for (filename, label), color in zip(files, colors):
    df = data_dict[label]
    plt.plot(df['Time_sec'], df['Abs'], label=label, linewidth=1.5, color=color)

plt.xlabel('Время (сек)', fontsize=12)
plt.ylabel('Поглощение D = lg(I₀/I)', fontsize=12)
plt.title('Зависимость поглощения от времени для всех образцов', fontsize=14)
plt.legend(loc='best', fontsize=10, ncol=2)
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()

# Дополнительный график: отдельные подграфики для каждого образца
fig, axes = plt.subplots(5, 2, figsize=(15, 18))
axes = axes.flatten()

for idx, (filename, label) in enumerate(files):
    df = data_dict[label]
    axes[idx].plot(df['Time_sec'], df['Abs'], linewidth=1.5, color=colors[idx])
    axes[idx].set_title(label, fontsize=12)
    axes[idx].set_xlabel('Время (сек)', fontsize=10)
    axes[idx].set_ylabel('Поглощение D', fontsize=10)
    axes[idx].grid(True, alpha=0.3)

plt.suptitle('Графики поглощения для каждого образца', fontsize=14)
plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

files = ['AION.csv', 'BION.csv', 'CION.csv', 'DION.csv', 'EION.csv']

plt.figure()

for file in files:
    # Чтение
    df = pd.read_csv(file, skiprows=1)
    df.columns = ['time', 'absorption']
    
    # Берем только первые 400 точек (если нужно)
    df = df.iloc[:400].copy()
    
    # Начальное значение D0
    D0 = df['absorption'].iloc[0]
    
    # Проверка (важно!)
    if D0 == 0:
        raise ValueError(f"В файле {file} D0 = 0, логарифм невозможен")
    
    # ln(D / D0)
    df['log_ratio'] = np.log(df['absorption'] / D0)
    
    # Подпись линии = имя файла без .csv
    label = file.replace('.csv', '')
    
    # График
    plt.plot(df['time'], df['log_ratio'], label=label)

plt.xlabel('Время, с')
plt.ylabel(r'$\ln\left(\frac{D}{D_0}\right)$')

plt.legend()
plt.grid()

plt.savefig("ALL_IONS.png")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

files = ['AION.csv', 'BION.csv', 'CION.csv', 'DION.csv', 'EION.csv']

# Ионные силы (в том же порядке)
ionic_strength = {
    'AION': 0.30,
    'BION': 0.60,
    'CION': 0.90,
    'DION': 1.70,
    'EION': 0.10
}

k_values = []
I_values = []

plt.figure()

for file in files:
    name = file.replace('.csv', '')
    I = ionic_strength[name]
    
    df = pd.read_csv(file, skiprows=1)
    df.columns = ['time', 'absorption']
    df = df.iloc[:400].copy()
    
    # фильтр (важно для логарифма)
    df = df[df['absorption'] > 0]
    
    D0 = df['absorption'].iloc[0]
    df['log_ratio'] = np.log(df['absorption'] / D0)
    
    # линейная аппроксимация: ln(D/D0) = k*t
    k, b = np.polyfit(df['time'], df['log_ratio'], 1)
    
    k_values.append(k)
    I_values.append(I)
    
    plt.plot(df['time'], df['log_ratio'], label=f'I = {I}')

plt.xlabel('Время, с')
plt.ylabel(r'$\ln\left(\frac{D}{D_0}\right)$')
plt.legend()
plt.grid()
plt.savefig("ALL_IONS.png")
plt.show()

# --- График зависимости ---
I_values = np.array(I_values)
k_values = np.array(k_values)

x = np.sqrt(I_values) / (1 + np.sqrt(I_values))
y = np.log(np.abs(k_values))  # модуль на случай отрицательных k

plt.figure()
plt.scatter(x, y)

# аппроксимация
a, b = np.polyfit(x, y, 1)
x_fit = np.linspace(min(x), max(x), 100)
y_fit = a * x_fit + b

plt.plot(x_fit, y_fit)

plt.xlabel(r'$\frac{\sqrt{I}}{1+\sqrt{I}}$')
plt.ylabel(r'$\ln k$')
plt.grid()

plt.savefig("lnk_vs_I.png")
plt.show()

print(a)
print(b)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

files = ['WCC1.csv', 'WCC2.csv', 'WCC3.csv', 'WCC4.csv', 'WCC5.csv']

# Концентрации NaOH (в том же порядке)
concentrations = [0.05, 0.1, 0.15, 0.2, 0.3]

# --- График 1: Поглощение ---
plt.figure()

for file, c in zip(files, concentrations):
    df = pd.read_csv(file, skiprows=1)
    df.columns = ['time', 'absorption']
    df = df.iloc[:450].copy()
    
    plt.plot(df['time'], df['absorption'], label=f'{c} M')

plt.xlabel('Время, с')
plt.ylabel(r'$D = \lg\left(\frac{I_0}{I}\right)$')
plt.title('Поглощение от времени')
plt.legend(title='[NaOH]')
plt.grid()

plt.savefig("WCC_absorption.png")
plt.show()


# --- График 2: ln(D/D0) ---
plt.figure()

for file, c in zip(files, concentrations):
    df = pd.read_csv(file, skiprows=1)
    df.columns = ['time', 'absorption']
    df = df.iloc[:450].copy()
    
    # фильтр для логарифма
    df = df[df['absorption'] > 0]
    
    D0 = df['absorption'].iloc[0]
    df['log_ratio'] = np.log(df['absorption'] / D0)
    
    plt.plot(df['time'], df['log_ratio'], label=f'{c} M')

plt.xlabel('Время, с')
plt.ylabel(r'$\ln\left(\frac{D}{D_0}\right)$')
plt.title('Логарифмическое изменение поглощения')
plt.legend(title='[NaOH]')
plt.grid()

plt.savefig("WCC_log.png")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

files = ['WCC1.csv', 'WCC2.csv', 'WCC3.csv', 'WCC4.csv', 'WCC5.csv']
concentrations = np.array([0.05, 0.1, 0.15, 0.2, 0.3])

k_values = []

for file in files:
    df = pd.read_csv(file, skiprows=1)
    df.columns = ['time', 'absorption']
    df = df.iloc[:400].copy()
    
    # убираем плохие значения
    df = df[df['absorption'] > 0]
    
    D0 = df['absorption'].iloc[0]
    df['log_ratio'] = np.log(df['absorption'] / D0)
    
    # линейная аппроксимация
    k, b = np.polyfit(df['time'], df['log_ratio'], 1)
    
    k_values.append(k)

k_values = np.array(k_values)

# Вывод коэффициентов
print("Угловые коэффициенты k:")
for c, k in zip(concentrations, k_values):
    print(f"C = {c} M: k = {k:.6e}")

# --- График ln(k) vs ln(C) ---
lnC = np.log(concentrations)
lnk = np.log(np.abs(k_values))  # берем модуль

plt.figure()
plt.scatter(lnC, lnk)

# аппроксимация (наклон = порядок реакции!)
a, b = np.polyfit(lnC, lnk, 1)
x_fit = np.linspace(min(lnC), max(lnC), 100)
y_fit = a * x_fit + b

plt.plot(x_fit, y_fit)

plt.xlabel(r'$\ln C_{\mathrm{NaOH}}$')
plt.ylabel(r'$\ln k$')
plt.title(f'Порядок реакции ≈ {a:.2f}')

plt.grid()
plt.savefig("lnk_vs_lnC.png")
plt.show()
print(a)
print(b)
```
